<h3>Task 1 (on paper)</h3>
<a id="task1"></a>

Show that $\ket{-}$ and $\ket{+}$ are eigenvectors for the $X$ operator with eigenvalues -1 and 1 respectively.

<h3>Solution </h3>



\begin{align*}
X \ket{-} = X \frac {\ket{0} - \ket{1}} {\sqrt{2}}
& =  \frac {\ket{1} - \ket{0}} {\sqrt{2}}\\
& = -\frac {\ket{0} - \ket{1}} {\sqrt{2}}\\
& = -\ket{-}
\end{align*}


\begin{align*}
X \ket{+} = X \frac {\ket{0} + \ket{1}} {\sqrt{2}}
& =  \frac {\ket{1} + \ket{0}} {\sqrt{2}}\\
& = \frac {\ket{0} + \ket{1}} {\sqrt{2}}\\
& = \ket{+}
\end{align*}

Hence $ \ket{-}$ and $\ket{+}$ are eigenvectors of $ X $ operator with the eigenvalues -1 and 1 respectively.

Consider the following quantum state where $ x=0 $ or $ 1 $. How can you find out the value of $x$ by applying a single operator?

$$
\frac {\ket{0} + (-1)^x \ket{1}} {\sqrt{2}}
$$


<h3>Solution </h3>

If our aim is to find out $ x $, then applying a Hadamard gate we can determine whether $ x=0 $ or $ 1 $ depending on the outcome. If the outcome is $ \ket{0} $, then $ x=0 $ and if the outcome is $ \ket{1} $, then $ x=1 $.

<h3>Task 3</h3>

Consider the unitary operator $U$ with eigenvector $\ket{1}$ and eigenvalue $e^{2\pi i \phi}$ where $\phi = \frac{5}{16}$. $CU$ operator can be realized as follows:

```python
from qiskit.circuit.library import CPhaseGate
from qiskit.circuit import Gate
from numpy import pi
import numpy as np

phase = 5/16
CU = CPhaseGate(theta=2*pi*phase)  # We need a 2\pi since CPhaseGate adds a phase of e^{i\theta}.
```

Pick $t=4$ and implement the phase estimation circuit to estimate $\phi$. Write a function named <i>qpe</i> which takes as parameters <i>t</i>, <i>control</i> qubits, <i>target</i> qubits, <i>circuit</i> and the operator <i>CU</i>.

Note that you will get an exact result since $t=4$ is precise enough.

<i>Remarks:</i>

- You can apply arithmetical operations to gates, such as taking powers or inverses.
- To apply $QFT^{\dagger}$, you can use the function you have written in the previous notebook.
- Some code is already given to guide you.


<h3>Solution </h3>

```python
#%%writefile qpe.py
from iqft import iqft

def qpe(t, control, target, circuit, CU):
    # Apply Hadamard to control qubits
    circuit.h(control)
    # Apply CU gates
    for i in range(t-1, -1, -1):
        # Obtain the power of CU gate
        CUi = CU**(2**(i))
        # Apply CUi gate where t-i-1 is the control
        qubits = [control[i]] + list(target)
        circuit.append(CUi, qubits)
    circuit.barrier()
    # Apply inverse QFT
    iqft(t, control, circuit)
```

```python
from qiskit import QuantumRegister, QuantumCircuit, ClassicalRegister
from qiskit.visualization import plot_histogram
from qiskit.circuit.library import CPhaseGate
from qiskit_aer import QasmSimulator
from numpy import pi

# Set n and t
n = 1
t = 4

# Create control and target qubits
control_reg = QuantumRegister(t, "c")
target_reg = QuantumRegister(n, "t")

# Create classical register to record measurement results on control qubits
cl_reg = ClassicalRegister(t, "cl")

# Create cirucit
circuit = QuantumCircuit(control_reg, target_reg, cl_reg)

# Set target qubit to store eigenvector
circuit.x(target_reg)
circuit.barrier()

# Apply QPE
qpe(t, control_reg, target_reg, circuit, CU)
circuit.barrier()

# Mesaure and print the results
circuit.measure(control_reg, cl_reg)
# display(circuit.draw("mpl"))

s = QasmSimulator()
print('Sample the circuit:')
counts = s.run(circuit, shots=1000).result().get_counts()
# Print a histogram of results
plot_histogram(counts)
```

<h3>Task 5</h3>

Try Task 3 this time by setting $\phi=0.685$. Calculate and print the estimates for different $t$ values between 1 and 10.

<h3>Solution </h3>

```python
from qiskit import QuantumRegister, QuantumCircuit, ClassicalRegister
from qiskit.visualization import plot_histogram
from qiskit_aer import QasmSimulator
from mufzk_quantum import bin2dec
import numpy as np

# Create CU operator with phase 0.685
phase = 0.685
CU = CPhaseGate(theta=2*pi*phase)

print("t\tphi")
print("-"*15)
for t in range (1, 11):
    control_reg = QuantumRegister(t, "cont")
    target_reg = QuantumRegister(n, "trgt")
    cl_reg = ClassicalRegister(t, "meas")

    qc = QuantumCircuit(control_reg, target_reg, cl_reg)

    qc.x(target_reg)
    qc.barrier()

    qpe(t, control_reg, target_reg, qc, CU)

    qc.measure(control_reg, cl_reg)

    sim = QasmSimulator()
    counts = sim.run(qc).result().get_counts()
    counts = dict(sorted(counts.items(), reverse=True, key=lambda key: key[1]))
    x_bin = list(counts.keys())[0]
    x_dec = bin2dec(x_bin)
    phi = x_dec / 2 ** t
    print(f"{t}\t{round(phi, 3)}")
```

<h3>Task 6</h3>

You are given an operator $U$ with eigenvector $\ket{11}$ and eigenvalue $\phi$, which is this time unknown to you. Use phase estimation to estimate $\phi$ by trying different $t$ values.

Run the following cell to load operator $CU$

```python
from mufzk_quantum import random_phase_unitary
from qiskit.circuit.library import UnitaryGate

n = 2
U = random_phase_unitary(num_qubits=n)
CU = UnitaryGate(data=U).control(num_ctrl_qubits=1)
```

<h3>Solution </h3>

```python
from qiskit import QuantumRegister, QuantumCircuit, ClassicalRegister
from qiskit_aer import QasmSimulator
from mufzk_quantum import bin2dec, get_phase
import numpy as np

# Original theta value
theta_org = get_phase(U)
# Original phi value
phi_org = theta_org / (2 * np.pi)
print("Original values:")
print(f"theta = {round(theta_org, 3)}, phi = {round(phi_org, 3)}")
print()
print("Obtained values:")
print("t\ttheta\tphi\t  PE")
print("-"*30)
for t in range (1, 11):
    control_reg = QuantumRegister(t, "cont")
    target_reg = QuantumRegister(n, "trgt")
    cl_reg = ClassicalRegister(t, "meas")

    qc = QuantumCircuit(control_reg, target_reg, cl_reg)

    qc.x(target_reg)
    qc.barrier()

    qpe(t, control_reg, target_reg, qc, CU)

    qc.measure(control_reg, cl_reg)

    sim = QasmSimulator()
    counts = sim.run(qc).result().get_counts()
    counts = dict(sorted(counts.items(), reverse=True, key=lambda key: key[1]))
    x_bin = list(counts.keys())[0]
    x_dec = bin2dec(x_bin)
    phi = x_dec / 2 ** t
    theta = 2 * np.pi * phi
    pe = str(round(abs(phi_org-phi)/phi*100)).rjust(3, ' ')
    print(f"{t}\t{round(theta, 3)}\t{round(phi, 3)}\t{pe} %")
```
