<h3>Task 1 (on paper)</h3>

Show that after applying the controlled $U^{2^j}$ gates for $j=0,\dots,t-1$, the state obtained can be expressed as $\displaystyle \frac{1}{2^{t/2}} \sum_{k=0}^{2^t-1} U^k\ket{\psi} \ket{k} $.

<h3>Solution</h3>

We have already seen the effect of the first gate corresponding to case $j=0$. Let's continue with the remaining gates.

- $ j=0 $, qubit $ t$ is the control.

If $ k_t=0 $, new state is  $\ket{\psi}\ket{k}$. If $ k_t=1 $, new state is $\ket{\psi} e^{2\pi i \phi 2^0} \ket{k} = e^{2\pi i \phi 2^0} \ket{\psi}\ket{k} $.

Hence, we can write it as, $U^{k_t 2 ^0} \ket{\psi}\;\ket{k} .$

- $ j=1 $, qubit $ t-1 $ is the control.

If $ k_{t-1}=0 $, new state is $ U^{k_t 2 ^0} \ket{\psi}\;\ket{k} $. If $ k_{t-1}=1 $, new state is $ e^{2\pi i \phi 2^1} U^{k_t 2 ^0} \ket{\psi}\;\ket{k} $.

Hence, we can write it as, $ U^{k_{t-1} 2 ^1} U^{k_t 2 ^0} \ket{\psi}\;\ket{k}$

Applying the $ CU^{2^j} $ gates for each qubit, we obtain the following state at the end.

$ U^{k_1 (2^t-1)} \cdots U^{k_{t-1} 2^1}   U^{k_t 2^0} \ket{\psi} \;\ket{k} = U^{k_1 (2^t-1)+\cdots +k_{t-1} 2^1+ k_t 2^0} \ket{\psi} \;\ket{k}
 $

Now noting that $ k_1 (2^t-1)+\cdots +k_{t-1} 2^1+k_t 2^0=k $, we can express the state obtained as $\displaystyle \frac{1}{2^{t/2}} \sum_{k=0}^{2^t-1} U^k\ket{\psi}\;\ket{k}$.

<h3>Solution</h3>

```python
from qiskit import QuantumRegister, QuantumCircuit
from qiskit.circuit.library import UnitaryGate
from qiskit_aer import StatevectorSimulator
from mufzk_quantum import print_statevector, order_finder_unitary

#Set t and n, sizes of the registers
t = 12
n = 4
L = 4

#Create control and target qubits
ctrl_reg = QuantumRegister(t, "1")
trgt_reg = QuantumRegister(L, "2")

#Create a circuit
circuit = QuantumCircuit(ctrl_reg, trgt_reg)

#Create the operator CU
x=7
N=15
U = order_finder_unitary(x, N)
CU = UnitaryGate(U).control(num_ctrl_qubits=1)

#Apply Hadamard to control qubits
circuit.x(trgt_reg[0])
circuit.h(ctrl_reg)

#Apply CU gates
for k in range(t-1, -1, -1):
    qubits = [ctrl_reg[k]] + list(trgt_reg)
    circuit.append(CU ** (2 ** k), qubits)
display(circuit.draw("mpl"))

#Simulate the circuit
sv_sim = StatevectorSimulator()
sv = sv_sim.run(circuit).result().get_statevector()

#Use Dirac notation to print the output
print_statevector(sv)
```

Let's check the first 5 states. The first 12 bits represent the first register and the last 4 bits represent the second register.

$\ket{0001}\ket{000000000000} = \ket{1}\ket{0}$

$\ket{0111}\ket{000000000001} = \ket{7}\ket{1}$

$\ket{0100}\ket{000000000010} = \ket{4}\ket{2}$

$\ket{1101}\ket{000000000011} = \ket{13}\ket{3}$

$\ket{0001}\ket{000000000100} = \ket{1}\ket{4}$

<h3>Task 3</h3>

Measure the second register and sample the circuit. Next, simulate the circuit and print the obtained state using <i>print_statevector()</i>. Check the first five states and convert to integer representation.

<h3>Solution</h3>

```python
from qiskit import ClassicalRegister
from qiskit.visualization import plot_histogram
from qiskit_aer import QasmSimulator

#Measure the target register
cr_trgt = ClassicalRegister(L)
circuit.add_register(cr_trgt)
circuit.measure(trgt_reg, cr_trgt)

#Sample the circuit
print('Sample the circuit:')
qasm_sim =  QasmSimulator()
job = qasm_sim.run(circuit)
counts = job.result().get_counts()
# Print a histogram of results
display(plot_histogram(counts))

#Simulate the circuit. One of the outcomes is picked due to measurement
sv = sv_sim.run(circuit).result().get_statevector()
print_statevector(sv)
```

In our case, the second register is measured as $\ket{1101}$ which is $\ket{13}$. Note that you might get a different measurement result. Let's check the first 5 states.

$\ket{000000000011}=\ket{3}$

$\ket{000000000111} = \ket{7}$

$\ket{000000001011} = \ket{11}$

$\ket{000000001111} = \ket{15}$

$\ket{000000010011} = \ket{19}$
